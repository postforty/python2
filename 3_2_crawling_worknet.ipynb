```python
# 봇 처럼 여겨지지 않기 위해 주피터 노트북 ipynb 파일 생성
# 크롤링을 어떻게 사이트에서 사람이 하는 것처럼 보일까가 중요
# pip install webdriver-manager
# 크롬 드라이버 파일을 손쉽게 다운로드
from webdriver_manager.chrome import ChromeDriverManager
# pip install selenium
from selenium import webdriver

driver = webdriver.Chrome(ChromeDriverManager().install())

srcKeyword = 'python'
resultCntInfo = '1000'
region = '26000'

# 워크넷 채용정보 상세 검색
URL=f'https://www.work.go.kr/empInfo/empInfoSrch/list/dtlEmpSrchList.do?careerTo=&keywordJobCd=&occupation=&templateInfo=&shsyWorkSecd=&rot2WorkYn=&payGbn=&resultCnt=10&keywordJobCont=&cert=&cloDateStdt=&moreCon=more&minPay=&codeDepth2Info=11000&isChkLocCall=&sortFieldInfo=DATE&major=&resrDutyExcYn=&eodwYn=&sortField=DATE&staArea=&sortOrderBy=DESC&keyword=&termSearchGbn=all&carrEssYns=&benefitSrchAndOr=O&disableEmpHopeGbn=&webIsOut=&actServExcYn=&maxPay=&keywordStaAreaNm=&emailApplyYn=&listCookieInfo=DTL&pageCode=&codeDepth1Info=11000&keywordEtcYn=&publDutyExcYn=&keywordJobCdSeqNo=&exJobsCd=&templateDepthNmInfo=&computerPreferential=&regDateStdt=&employGbn=&empTpGbcd=1&region={region}&infaYn=&resultCntInfo={resultCntInfo}&siteClcd=all&cloDateEndt=&sortOrderByInfo=DESC&currntPageNo=1&indArea=&careerTypes=&searchOn=Y&tlmgYn=&subEmpHopeYn=&academicGbn=&templateDepthNoInfo=&foriegn=&mealOfferClcd=&station=&moerButtonYn=&holidayGbn=&srcKeyword={srcKeyword}&enterPriseGbn=all&academicGbnoEdu=noEdu&cloTermSearchGbn=all&keywordWantedTitle=&stationNm=&benefitGbn=&keywordFlag=&notSrcKeyword=&essCertChk=&isEmptyHeader=&depth2SelCode=&_csrf=d0fab365-9f79-4601-8f1d-c28beb63f8cb&keywordBusiNm=&preferentialGbn=&rot3WorkYn=&pfMatterPreferential=&regDateEndt=&staAreaLineInfo1=11000&staAreaLineInfo2=1&pageIndex=1&termContractMmcnt=&careerFrom=&laborHrShortYn=#viewSPL'
driver.get(url=URL)

driver.implicitly_wait(time_to_wait=10)
```

```python
from selenium.webdriver.common.by import By

result = []
conpany = driver.find_elements(By.CSS_SELECTOR, "a.cp_name")
title = driver.find_elements(By.CSS_SELECTOR, "div.cp-info-in > a")
pay = driver.find_elements(By.XPATH, "//tr/td[4]/div/p[1]") # 크롬 개발자 모드에서 copy
dday = driver.find_elements(By.CSS_SELECTOR, "div.cp-info > p.dday.mt10")
address = driver.find_elements(By.XPATH, "//td[3]/div/p[2]/em[3]") # 크롬 개발자 모드에서 copy
href = driver.find_elements(By.CSS_SELECTOR, "div.cp-info-in > a")

# print(images.get_attribute('src'))
for i in range(len(conpany)):
    temp = []
    # 회사
    if conpany[i].get_attribute('innerText') is not None:
        temp.append(conpany[i].get_attribute('innerText'))
    else:
        temp.append("")
    # 공고 제목
    if title[i].get_attribute('innerText') is not None:
        temp.append(title[i].get_attribute('innerText'))
    else:
        temp.append("")
    # 연봉
    if pay[i].get_attribute('innerText') is not None:
        temp.append(pay[i].get_attribute('innerText'))
    else:
        temp.append("")
    # 채용기간
    if dday[i].get_attribute('innerText') is not None:
        temp.append(dday[i].get_attribute('innerText'))
    else:
        temp.append("")
    # 주소
    if address[i].get_attribute('innerText') is not None:
        temp.append(address[i].get_attribute('innerText'))
    else:
        temp.append("")
    # 공고 링크
    if href[i].get_attribute('innerText') is not None:
        temp.append(href[i].get_attribute('href'))
    else:
        temp.append("")
        
    result.append(temp)

print(' 찾은 이미지 개수: ', len(result))
# for i in result:
#     print(i)
```

```python
# pip install openpyxl
import pandas as pd
from datetime import datetime

index_list = list(range(1, len(result)+1))
df = pd.DataFrame(result, columns =["회사", "채용공고명", "연봉", "마감일", "위치", "링크"], index=index_list)

print(df)

# datetime.today().strftime("%Y%m%d%H%M%S")    # YYYYmmddHHMMSS 형태의 시간 출력
# datetime.today().strftime("%Y/%m/%d %H:%M:%S")  # YYYY/mm/dd HH:MM:SS 형태의 시간 출력

today = datetime.today().strftime("%y%m%d")
# df.to_excel(f'worknet_{srcKeyword}_{region}_{today}.xlsx', index=True, header=True)
df.to_excel(f'worknet_{srcKeyword}_{region}_{today}.xlsx', index=True, header=True)


# https://wikidocs.net/151203
# 엑셀 컬럼 자동 맞춤
import win32com.client
# 엑셀 프로그램 실행
excel = win32com.client.Dispatch("Excel.Application")
# 실행된 엑셀의 시각화
excel.Visible = False

import os

xlsx = os.path.abspath(f'worknet_{srcKeyword}_{region}_{today}.xlsx')

# 기존 파일 열기
wb = excel.Workbooks.Open(xlsx)
ws = wb.ActiveSheet

# 모든 컬럼 자동 맞추기
ws.Columns.AutoFit()

# 저장
wb.Save()

# 파일 종료
wb.Close()

# excel 종료
excel.Quit()
```
