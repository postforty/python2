```python
import requests
from bs4 import BeautifulSoup

# HTML 파일 읽기
response = requests.get("http://127.0.0.1:5500/sample.html")
response
```

```python
soup = BeautifulSoup(response.content, 'html.parser')
print(soup.prettify())
```

```python
# 첫 번째 h1 태그 찾기
h1_tag = soup.find('h1')
print("제목:", h1_tag.text)
```

```python
# 첫 번째 도서 제목 찾기
first_title = soup.find('h2', class_='title')
print("첫 번째 책:", first_title.text)
```

```python
# 모든 도서 정보 찾기
books = soup.find_all('div', class_='book')

print(f"총 {len(books)}권의 책이 있습니다.\n")
```

```python
# ID로 특정 도서 찾기
book2 = soup.find('div', id='book2')
book2
```

```python
# 1. find() - 첫 번째 요소 찾기
element = soup.find('div', class_='book')
element
```

```python
# 2. find_all() - 모든 요소 찾기
elements = soup.find_all('p')
elements
```

```python
# 3. select() - CSS 선택자 사용
elements = soup.select('.book .title')  # class가 book인 요소 안의 title
elements
```

```python
# 4. select_one() - CSS 선택자로 첫 번째 요소 찾기
element = soup.select_one('#book1')
element
```

```python
# 5. get() - 속성 값 가져오기
link = soup.find('a')
href = link.get('href')  # 또는 link['href']

href
```
