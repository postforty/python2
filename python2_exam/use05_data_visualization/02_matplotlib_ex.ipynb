```python
plt.plot([1, 2, 3, 4], [2, 5, 3, 6], label='this year') # x-y 값
plt.xlabel('X-Label') # 레이블 표시
plt.ylabel('Y-Label')
plt.legend() # * 범례 표시
# plt.legend(loc='lower right') # * 범례 표시 + 위치 설정
plt.show()
```

```python
plt.plot([1, 2, 3, 4], [2, 5, 3, 6], label='this year')
plt.plot([1, 2, 3, 4], [5, 2, 6, 3], label='last year')
plt.xlabel('X-Axis')
plt.ylabel('Y-Axis')
# plt.legend(loc='best')
plt.legend(loc='best', ncol=2)

plt.show()
```

### 🎨 Matplotlib 시각화 설정 단축 코드 정리

#### 1. 색상 (Colors)

| Character | Color |
| :---: | :--- |
| `b` | blue (파란색) |
| `g` | green (초록색) |
| `r` | red (빨간색) |
| `c` | cyan (청록색) |
| `m` | magenta (자홍색) |
| `y` | yellow (노란색) |
| `k` | black (검은색) |
| `w` | white (흰색) |

#### 2. 선 스타일 (Line Styles)

| Character | Description |
| :---: | :--- |
| `-` | solid line style (실선 스타일) |
| `--` | dashed line style (대시선 스타일) |
| `-.` | dash-dot line style (대시-점선 스타일) |
| `:` | dotted line style (점선 스타일) |

#### 3. 마커 (Markers)

| Character | Description |
| :---: | :--- |
| `.` | point marker (점 마커) |
| `,` | pixel marker (픽셀 마커) |
| `o` | circle marker (원 마커) |
| `v` | triangle\_down marker (역삼각형 마커) |
| `^` | triangle\_up marker (삼각형 마커) |
| `<` | triangle\_left marker (왼쪽 삼각형 마커) |
| `>` | triangle\_right marker (오른쪽 삼각형 마커) |
| `1` | tri\_down marker (아래쪽 삼각 마커) |
| `2` | tri\_up marker (위쪽 삼각 마커) |
| `3` | tri\_left marker (왼쪽 삼각 마커) |
| `4` | tri\_right marker (오른쪽 삼각 마커) |
| `s` | square marker (정사각형 마커) |
| `p` | pentagon marker (오각형 마커) |
| `*` | star marker (별 마커) |
| `h` | hexagon1 marker (육각형1 마커) |
| `H` | hexagon2 marker (육각형2 마커) |
| `+` | plus marker (플러스 마커) |
| `x` | x marker (엑스 마커) |
| `D` | diamond marker (다이아몬드 마커) |
| `d` | thin\_diamond marker (얇은 다이아몬드 마커) |
| `\|` | vline marker (수직선 마커) |
| `_` | hline marker (수평선 마커) |

```python
import matplotlib.pyplot as plt
import numpy as np

years = ['2022', '2023', '2024', '2025']
sales = [2, 5, 3, 6]      # 판매량
profit = [1.5, 4, 2.5, 5] # 수익

x = np.arange(len(years))  # [0, 1, 2, 3]
width = 0.35  # 막대 너비

# 2개의 막대 그래프
plt.bar(x - width/2, sales, width, label='판매량', color='skyblue')
plt.bar(x + width/2, profit, width, label='수익', color='orange')

plt.xlabel('년도')
plt.ylabel('값')
plt.title('년도별 판매량과 수익')
plt.xticks(x, years)  # x축 레이블
plt.legend()  # 범례 표시

plt.show()
```

데이터프렘을을 막대 그래프로 시각화

```python
import matplotlib.pyplot as plt
import pandas as pd

# 데이터프레임 생성
data = {
    '년도': ['2022', '2023', '2024', '2025'],
    '판매량': [2, 5, 3, 6],
    '수익': [1.5, 4, 2.5, 5]
}
df = pd.DataFrame(data)

# 년도를 인덱스로 설정
df = df.set_index('년도')

# 막대 그래프
df.plot(kind='bar', color=['skyblue', 'orange'], width=0.8)

plt.xlabel('년도')
plt.ylabel('값')
plt.title('년도별 판매량과 수익')
plt.xticks(rotation=0)
plt.legend()
plt.show()
```

```python
import matplotlib.pyplot as plt
import pandas as pd

# 데이터프레임 생성
data = {
    '년도': ['2022', '2023', '2024', '2025'],
    '판매량': [2, 5, 3, 6],
    '수익': [1.5, 4, 2.5, 5]
}
df = pd.DataFrame(data)
df = df.set_index('년도')

# figure와 axes 생성
fig, ax = plt.subplots()

# 막대 그래프
df.plot(kind='bar', ax=ax, color=['skyblue', 'orange'], width=0.8)

ax.set_xlabel('년도')
ax.set_ylabel('값')
ax.set_title('년도별 판매량과 수익')
ax.tick_params(axis='x', rotation=0)

plt.show()
```
