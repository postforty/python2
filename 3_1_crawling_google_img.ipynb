```python
# 봇 처럼 여겨지지 않기 위해 주피터 노트북 ipynb 파일 생성
# 크롤링을 어떻게 사이트에서 사람이 하는 것처럼 보일까가 중요
# pip install webdriver-manager
# 크롬 드라이버 파일을 손쉽게 다운로드
from webdriver_manager.chrome import ChromeDriverManager
# pip install selenium
from selenium import webdriver

driver = webdriver.Chrome(ChromeDriverManager().install())

URL='https://www.google.co.kr/imghp'
driver.get(url=URL)

driver.implicitly_wait(time_to_wait=10)
```

```python
# 키 입력을 위한 라이브러리
from selenium.webdriver.common.keys import Keys
from selenium.webdriver.common.by import By

elem = driver.find_element(By.CSS_SELECTOR, "textarea.gLFyf") # "태그.클래스선택자#아이디선택자 > 태그"
elem.send_keys("바다")
elem.send_keys(Keys.RETURN)
```

```python
import time
elem = driver.find_element(By.CSS_SELECTOR, "body")
for i in range(60):
    elem.send_keys(Keys.PAGE_DOWN)
    time.sleep(0.1)

try:
    driver.find_element(By.CSS_SELECTOR, "input.mye4qd").click() # 결과 더보기 버튼이 있는 경우 클릭!

    for i in range(60):
        elem.send_keys(Keys.PAGE_DOWN)
        time.sleep(0.1)
except:
    pass

```

```python
links = []
images = driver.find_elements(By.CSS_SELECTOR, "a.wXeWr.islib.nfEiy > div.bRMDJf.islir > img") #find_elements 복수개
# print(images.get_attribute('src'))
for image in images:
    if image.get_attribute('src') is not None:
        links.append(image.get_attribute('src'))

print(' 찾은 이미지 개수: ', len(links))
```

```python
import urllib.request

count = 0
for idx, value in enumerate(links):
    url = value
    urllib.request.urlretrieve(url, f'./crawling_img/bada_{idx}.jpg')
    count += 1

print(f'{count}개 이미지 다운로드 완료')

# ipynb를 py로~
# jupyter nbconvert --to script 3_1_crawling.ipynb 

```
