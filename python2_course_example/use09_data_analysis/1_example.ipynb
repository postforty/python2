```python
# 히스트그램(수치형 데이터인 경우 사용)
raw['분양가격(제곱미터)_수치형'].hist(bins=30) # bins : 구간(막대) 개수
```


raw['분양가격(제곱미터)_수치형'].hist(bins=30)은 파이썬 판다스(Pandas) 라이브러리를 사용하여 "분양가격(제곱미터)_수치형" 열의 값을 기반으로 히스토그램을 생성하는 코드입니다. 히스토그램은 데이터의 분포를 시각적으로 나타내는 그래프로, 특정 범위에 속하는 값의 빈도를 막대(bar)로 표현합니다.

히스토그램 읽는 방법
히스토그램을 읽을 때, x축과 y축의 의미를 이해하는 것이 중요합니다.

1. X축 (가로축)
의미: x축은 "분양가격(제곱미터)_수치형"의 값 범위를 나타냅니다. 즉, 이 히스토그램에서 x축은 제곱미터당 분양 가격을 나타냅니다.
범위: bins=30 옵션에 의해 x축의 값이 30개의 구간(bin)으로 나뉩니다. 각 bin은 데이터 값의 특정 범위를 나타내며, 구간의 크기는 데이터의 전체 범위에 따라 자동으로 결정됩니다. 예를 들어, 분양가격이 5000에서 8000 사이에 걸쳐 있다면, 각 bin은 이 범위를 30개의 구간으로 나눈 작은 범위를 나타냅니다.
구간: 각 bin은 특정 분양가격 범위를 나타내며, 이 범위 안에 해당하는 데이터가 얼마나 많은지를 보여줍니다.
2. Y축 (세로축)
의미: y축은 각 bin에 속하는 데이터 포인트의 개수(빈도수)를 나타냅니다.
빈도수: 각 막대의 높이는 그 구간(bin)에 속하는 "분양가격(제곱미터)_수치형" 값의 개수를 의미합니다. 예를 들어, x축의 첫 번째 bin이 5000에서 5100까지의 분양가격 범위를 나타낸다면, y축은 이 범위에 속하는 데이터 포인트(아파트 가격)의 개수를 보여줍니다.
