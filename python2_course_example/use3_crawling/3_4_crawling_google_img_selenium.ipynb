# selenium을 이용한 크롤링


```python
# 봇 처럼 여겨지지 않기 위해 주피터 노트북 ipynb 파일 생성
# 크롤링을 어떻게 사이트에서 사람이 하는 것처럼 보일까가 중요

# pip install selenium
from selenium import webdriver

driver = webdriver.Chrome()

URL='https://www.google.co.kr/imghp'
driver.get(url=URL)
```

```python
# 키 입력을 위한 라이브러리
from selenium.webdriver.common.by import By
from selenium.webdriver.common.keys import Keys

elem = driver.find_element(By.CSS_SELECTOR, "textarea.gLFyf") # "태그.클래스선택자#아이디선택자 > 태그"
# elem.send_keys("고양이") # TODO 수정할 것
elem.send_keys("개") # TODO 수정할 것
elem.send_keys(Keys.RETURN)
```

```python
import time
elem = driver.find_element(By.CSS_SELECTOR, "body")
for i in range(100):
    elem.send_keys(Keys.PAGE_DOWN)
    time.sleep(0.5)

try:
    driver.find_element(By.CSS_SELECTOR, "input.LZ4I").click() # 결과 더보기 버튼이 있는 경우 클릭!

    for i in range(100):
        elem.send_keys(Keys.PAGE_DOWN)
        time.sleep(0.5)

except:
    pass

```

```python
links = []
images = driver.find_elements(By.CSS_SELECTOR, "div.H8Rx8c > g-img.mNsIhb > img") #find_elements 복수개
# print(images.get_attribute('src'))
for image in images:
    if image.get_attribute('src') is not None:
        links.append(image.get_attribute('src'))

print(' 찾은 이미지 개수: ', len(links))
```

```python
import urllib.request
# import os

# path = "./cat"
# try:
#     os.mkdir(path)
# except:
#     pass

from tqdm import tqdm

count = 0
for idx, value in enumerate(tqdm(links)):
    url = value
    urllib.request.urlretrieve(url, f'./dog/dog_{idx}.jpg') # 이미지의 주소와 다운받을 경로를 적어주면 해당 위치에 이미지를 내려 받는다.
    count += 1

print(f'{count}개 이미지 다운로드 완료')

```
